## 1. Python Output — `print()`

`print()` writes to the screen. It takes any number of values.

````python
print('Hello, world!')
````

````python
# multiple values — separated by a space by default
print('Data', 'Science', 'Mentorship', 'Program', 2026)
````

````python
# Python is CASE-SENSITIVE: print works, Print does not exist.
# Print('hello')  # ❌ NameError: name 'Print' is not defined
print('python is case sensitive')
````

````python
# ❌ text without quotes is code, not a string:
# two bare words → SyntaxError (one word would be a NameError)
print(salman khan)
````

### `sep` — what goes *between* values

````python
print('01', '01', '2026', sep='/')   # date-style
print('a', 'b', 'c', sep='---')
````

### `end` — what goes *after* the line (default `'\n'`)

````python
print('hello', end=' ')
print('world')            # continues on the same line

print('no newline', end='')
print('...glued')
````

## 3. Keywords & Identifiers

**Keywords** are reserved words — Python's own vocabulary. You cannot use them as names.

````python
import keyword

print(len(keyword.kwlist), 'keywords')
print(keyword.kwlist)
````

**Identifiers** are the names *you* create (variables, functions, ...).

Rules:
1. Start with a letter or `_` — never a digit
2. Only letters, digits, `_` (no spaces, no `-`, no `$`)
3. Not a keyword
4. Case-sensitive: `name` ≠ `Name`

````python
name1 = 'valid'          # ✅ digit allowed after first char
_hidden = 'valid'        # ✅ underscore start is fine
first_name = 'valid'     # ✅ snake_case — the Python convention
print(name1, _hidden, first_name)
````

````python
# ❌ identifier cannot start with a digit → SyntaxError
1name = 'invalid'
````

````python
# `_` alone is a valid identifier — often used for "value I don't care about"
_ = 'ignored'
print(_)
````

````python
# ⚠️ gotcha: you CAN name a variable after a built-in — and break it
sum = 0                    # shadows the built-in sum()
# sum([1, 2, 3])           # would now raise TypeError: 'int' not callable
del sum                    # remove the shadow → built-in visible again
print(sum([1, 2, 3]))
````

## 4. Data Types

```
                    Python data types
                           │
       ┌──────────┬────────┴───────┬─────────────┐
    numeric     text          collections      special
       │          │                │              │
  int float    str        list tuple set dict   None bool*
  complex

  *bool is technically a subclass of int (True == 1, False == 0)
```

`type()` reveals the type of any value.

````python
# numeric
print(8, type(8))                    # int
print(8.55, type(8.55))              # float
print(2 + 3j, type(2 + 3j))          # complex
````

````python
# int has ARBITRARY precision — no overflow, unlike C/Java
print(10 ** 100)
````

````python
# float is 64-bit — it DOES have a limit; beyond ~1.8e308 it becomes inf
print(1e308)
print(1e309)
````

````python
# inf and nan also exist as explicit values
print(float('inf'), float('-inf'))
print(float('nan'))              # not-a-number; nan != nan!
````

````python
# floats are BINARY fractions — tiny precision errors are normal
print(0.1 + 0.2)             # 0.30000000000000004
print(0.1 + 0.2 == 0.3)      # False!
print(round(0.1 + 0.2, 2))   # round() to compare/display
print(round(3.14159, 2), round(2.5), round(3.5))   # banker's rounding: to even
````

````python
# boolean
print(True, type(True))
print(False, type(False))
````

````python
# text
print('Hello', type('Hello'))
````

````python
# collections
print([1, 2, 3], type([1, 2, 3]))                          # list  — mutable
print((1, 2, 3), type((1, 2, 3)))                          # tuple — immutable
print({1, 2, 3}, type({1, 2, 3}))                          # set   — unique items
print({'name': 'Lakshay', 'age': 25}, type({'a': 1}))      # dict  — key:value
````

````python
# special: absence of a value
print(None, type(None))
````

## 5. Variables

A variable is a **name bound to an object** — a label stuck on a value, not a box
holding it.

```
 a = 5        name 'a' ──► object 5
 a = 'five'   name 'a' ──► object 'five'   (label moved; 5 unchanged)
```

````python
name = 'python'
print(name)

a = 5
b = 6
print(a + b)
````

### Dynamic typing vs static typing

| | Static (C, Java) | Dynamic (Python) |
|---|---|---|
| Declare type? | `int a = 5;` | `a = 5` |
| Type lives on | the **variable** | the **object** |
| Can type change? | no | yes — rebind to anything |

````python
a = 5
print(a, type(a))

a = 'now a string'   # same name, new type — dynamic typing
print(a, type(a))
````

### Dynamic binding

The name→object link is created **at runtime**, and can be re-created any time.
Static-language variables are bound to one type at compile time.

````python
a = 5          # bound to an int object
a = [1, 2, 3]  # re-bound to a list object at runtime
print(a)
````

### Multiple assignment — stylish declarations

````python
# unpack three values in one line
a, b, c = 1, 2, 3
print(a, b, c)
````

````python
# chain: all three names point to the SAME object
a = b = c = 5
print(a, b, c)
````

````python
# long lines: implicit continuation inside () [] {} — the clean way
total = (1 + 2 + 3 +
         4 + 5)
# ...or a backslash (works, but brittle)
also = 1 + 2 + \
       3
# semicolons chain statements on one line (legal, discouraged)
x = 1; y = 2
print(total, also, x + y)
````

## 6. User Input — `input()`

⚠️ `input()` **always returns a `str`**, whatever the user types.

````python
# In Colab/Jupyter, uncomment to type live:
# first = input('enter first number: ')
first = '15'    # sample value standing in for user input
second = '27'

print(type(first), type(second))
print(first + second)   # '1527' — string concatenation, not addition!
````

## 7. Type Conversion

**Implicit** — Python promotes types automatically when safe:

````python
print(5 + 5.6, type(5 + 5.6))    # int + float → float
print(True + 4)                  # bool behaves as 1 → 5
````

````python
# Implicit conversion refuses unsafe mixes:
# ❌ int + str → TypeError
4 + '4'
````

**Explicit** — you convert with `int()`, `float()`, `str()`, ...

````python
first = '15'
second = '27'
print(int(first) + int(second))   # now real addition → 42
````

````python
print(int(4.99))      # truncates toward zero → 4
print(float(4))       # 4.0
print(str(42), type(str(42)))
print(int('100'))     # numeric string → fine
````

````python
# ❌ non-numeric string cannot become int → ValueError
int('hello')
````

## 8. Literals

A literal is a raw value written directly in source code.

### Numeric literals — four bases

````python
a = 0b1010     # binary
b = 0o310      # octal
c = 0x12c      # hexadecimal
d = 100        # decimal

print(a, b, c, d)
````

````python
# going the other way: int → base string, and string-in-base → int
print(bin(10), oct(200), hex(300))
print(int('1010', 2), int('310', 8), int('12c', 16))
````

````python
# float literals — decimal and scientific notation
f1 = 10.5
f2 = 1.5e2     # 1.5 × 10²
f3 = 1.5e-3    # 1.5 × 10⁻³
print(f1, f2, f3)
````

````python
# complex literal — j is the imaginary unit
z = 3.14j
print(z, z.real, z.imag)
````

### Escape sequences — special characters inside strings

| Escape | Meaning |
|---|---|
| `\n` | newline |
| `\t` | tab |
| `\\` | one backslash |
| `\'` `\"` | literal quote |
| `\U0001f600` | unicode by code point |

````python
print('line1\nline2')
print('col1\tcol2')
print('back\\slash')
print('it\'s fine')
````

### String literals

````python
s1 = 'single quotes'
s2 = "double quotes"
s3 = '''triple quotes
span multiple
lines'''
s4 = "unicode: \U0001f600 \U0001F60D"
s5 = r"raw string: \n stays as backslash-n"
s6 = u"unicode prefix (legacy — every str is unicode in Python 3)"

print(s1)
print(s2)
print(s3)
print(s4)
print(s5)
print(s6)
````

````python
# no separate char type — a single character is just a 1-length str
ch = 'C'
print(ch, type(ch), len(ch))
````

### Boolean & special literals

````python
# True/False are just 1/0 in arithmetic
print(True + 4)    # 5
print(False + 10)  # 10
````

````python
# None — 'no value yet'; common placeholder before real assignment
k = None
print(k)
````
