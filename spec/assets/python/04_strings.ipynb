# 04 · Strings

**TL;DR** — a string is an **immutable sequence of Unicode characters**.
You can read any slice of it, but never change it in place.

```
 index ruler for  s = 'python'

    +---+---+---+---+---+---+
    | p | y | t | h | o | n |
    +---+---+---+---+---+---+
      0   1   2   3   4   5     positive
     -6  -5  -4  -3  -2  -1     negative
```

**Agenda**: create → index → slice → immutability → operations → methods →
formatting (f-strings) → practice programs

## 1. Creating Strings

````python
s1 = 'hello'
s2 = "hello"
s3 = '''multi
line'''
s4 = str(3.14)          # convert anything to str

print(s1, s2, s3.replace('\n', ' | '), s4)
````

````python
# quotes inside quotes — mix them or escape
print("it's python")
print('she said "hi"')
print('it\'s escaped')
````

## 2. Indexing — positive & negative

````python
s = 'python'
print(s[0])     # p
print(s[5])     # n
print(s[-1])    # n — last char
print(s[-6])    # p — first char
````

````python
# ❌ index past the end → IndexError
s = 'python'
s[6]
````

## 3. Slicing — `s[start : stop : step]`

`stop` excluded. Omitted pieces default to "from the edge".

````python
s = 'hello world'
print(s[0:5])       # hello
print(s[6:])        # world
print(s[:5])        # hello
print(s[:])         # whole copy
print(s[0:8:2])     # every 2nd char: 'hlow'
````

````python
# negative step walks BACKWARD
s = 'hello world'
print(s[::-1])        # classic reverse
print(s[-1:-6:-1])    # 'dlrow' — last five, reversed
print(s[8:0:-2])
````

## 4. Immutability

Strings cannot be edited or partially deleted — only replaced wholesale.

````python
# ❌ item assignment → TypeError
s = 'hello world'
s[0] = 'H'
````

````python
# ❌ deleting one char → TypeError
s = 'hello world'
del s[-1]
````

````python
# deleting the WHOLE name is fine — that unbinds the variable
s = 'hello world'
del s
# print(s)  # would now raise NameError
````

## 5. Operations on Strings

````python
# arithmetic: + concatenates, * repeats
print('delhi' + ' ' + 'mumbai')
print('delhi ' * 3)
print('*' * 30)
````

````python
# precedence applies to strings too: * binds tighter than +
print('a' + 'b' * 3)      # 'abbb', not 'ababab'
print(('a' + 'b') * 3)    # 'ababab'
````

````python
# relational: lexicographic (dictionary) order, char by char
print('delhi' == 'delhi')
print('mumbai' > 'pune')     # 'm' < 'p' → False
print('Pune' > 'pune')       # uppercase codes < lowercase → False
print('goa' > 'go')          # longer wins if prefix equal
````

````python
# logical: '' is falsy, non-empty is truthy
print(bool(''), bool('hello'))
print('hello' and 'world')   # 'world' — and returns last truthy
print('' or 'fallback')      # 'fallback'
````

````python
# loops & membership
for ch in 'delhi':
    print(ch, end=' ')
print()
print('D' in 'Delhi', 'd' in 'Delhi')   # case-sensitive
````

## 6. Common Functions & Methods

### Built-in functions that accept strings

````python
s = 'hello world'
print(len(s))
print(max(s))               # highest code point
print(min(s))               # ' ' — space is lowest here
print(sorted(s))            # list of chars, ascending
print(sorted(s, reverse=True))
````

### Case methods

````python
s = 'hello world'
print(s.capitalize())   # Hello world
print(s.title())        # Hello World
print(s.upper())        # HELLO WORLD
print('HELLO'.lower())  # hello
print('HeLLo'.swapcase())  # hEllO
````

````python
# methods return NEW strings — original untouched (immutability!)
s = 'hello'
s.upper()
print(s)   # still 'hello'
````

### Search: `count`, `find`, `index`, `startswith`, `endswith`

| | not found → |
|---|---|
| `find` | returns `-1` |
| `index` | raises `ValueError` |

````python
s = 'my name is lakshay'
print(s.count('a'))
print(s.find('name'))     # start index
print(s.find('xyz'))      # -1
print(s.index('name'))
print(s.startswith('my'))
print(s.endswith('shay'))
````

````python
# find/count accept start (and end) bounds — search a window
s = 'banana'
print(s.find('a'))         # 1 — first match
print(s.find('a', 2))      # 3 — first match at/after index 2
print(s.count('a', 0, 4))  # 2 — only within [0, 4)
````

````python
# ❌ index() on a missing substring → ValueError
'my name is lakshay'.index('xyz')
````

### Predicates: `isalnum`, `isalpha`, `isdigit`, `isidentifier`

````python
print('nitish1988'.isalnum())   # letters+digits only
print('nitish'.isalpha())
print('1988'.isdigit())
print('first_name'.isidentifier())
print('1name'.isidentifier())
print('HELLO'.isupper(), 'hello'.islower(), '   '.isspace())
````

### Split / Join / Replace / Strip

````python
s = 'who is the pm of india'
words = s.split()               # default: whitespace
print(words)
print(s.split('is'))
````

````python
print(' '.join(['who', 'is', 'the', 'pm']))
print('-'.join(['2026', '07', '29']))
````

````python
print('hi my name is lakshay'.replace('lakshay', 'goyal'))
print('aaaa'.replace('a', 'b', 2))     # limit replacements
````

````python
print('   padded   '.strip())
print('xxhelloxx'.strip('x'))
print('   left-only'.lstrip(), '|', 'right-only   '.rstrip(), '|')
````

````python
# partition — split ONCE around a separator → 3-tuple
email = 'lakshay.goyal@testsigma.com'
print(email.partition('@'))
````

### Padding & alignment: `center`, `ljust`, `rjust`, `zfill`

````python
print('menu'.center(20, '-'))
print('item'.ljust(10, '.') + '₹99')
print('42'.rjust(8))
print('7'.zfill(3))          # zero-pad numbers → '007'
````

### Characters ⇄ code points: `ord` / `chr`

Every character IS a number underneath (its Unicode code point) —
this is why `'a' < 'b'` works.

````python
print(ord('A'), ord('a'), ord('😀'))
print(chr(65), chr(97), chr(128512))
````

````python
# Caesar shift — ord/chr in action
word = 'abc'
print(''.join(chr(ord(ch) + 1) for ch in word))
````

## 7. String Formatting

Three generations — **f-strings win**.

````python
name, age = 'Lakshay', 25

# 1) % formatting (legacy)
print('my name is %s and age %d' % (name, age))

# 2) .format() with positional indices
print('hi my name is {} and age {}'.format(name, age))
print('hi my age is {1} and name {0}'.format(name, age))

# 3) f-string (3.6+) — expression inline, best readability
print(f'my name is {name} and age {age}')
````

````python
# f-strings evaluate any expression, and control precision
pi = 3.14159265
print(f'{pi:.2f}')          # 2 decimals
print(f'{10 ** 6:,}')       # thousands separator
print(f'{0.375:.1%}')       # percentage
print(f'{name=}')           # debug form: name='Lakshay'
````

````python
# alignment & padding — the modern ljust/rjust/center/zfill
item, price = 'mouse', 899.5
print(f'|{item:<10}|')      # left-align in width 10
print(f'|{item:>10}|')      # right-align
print(f'|{item:^10}|')      # center
print(f'|{item:.^10}|')     # custom fill char
print(f'{price:08.2f}')     # zero-pad to width 8
````

## 8. Practice Programs

````python
# length without len()
s = 'hello world'
count = 0
for ch in s:
    count += 1
print(count)
````

````python
# extract username from email
# email = input('email: ')
email = 'lakshay.goyal@testsigma.com'
at = email.find('@')
print(email[:at])
````

````python
# frequency of a character
s = 'hello how are you'
target = 'h'
freq = 0
for ch in s:
    if ch == target:
        freq += 1
print(freq)
````

````python
# remove a given character
s = 'nitish'
target = 'i'
result = ''
for ch in s:
    if ch != target:
        result += ch
print(result)
````

````python
# palindrome check
s = 'malayalam'
print('palindrome' if s == s[::-1] else 'not palindrome')
````

````python
# ...and without slicing
s = 'malayalam'
is_pal = True
for i in range(len(s) // 2):
    if s[i] != s[len(s) - 1 - i]:
        is_pal = False
        break
print(is_pal)
````

````python
# count words without split()
s = 'hi how are you doing'
words = 1
for ch in s:
    if ch == ' ':
        words += 1
print(words)
````

````python
# stronger version: BUILD the word list yourself (split() by hand)
s = 'hi how are you doing'
words = []
current = ''
for ch in s:
    if ch == ' ':
        words.append(current)
        current = ''
    else:
        current += ch
words.append(current)          # flush the last word
print(words, len(words))
````

````python
# title-case without title()
s = 'hi how are you'
result = ''
for word in s.split():
    result += word[0].upper() + word[1:] + ' '
print(result.strip())
````

````python
# int → str manually (without str())
number = 5382
digits = '0123456789'
result = ''
while number > 0:
    result = digits[number % 10] + result
    number //= 10
print(result, type(result))
````

## Recap

| Concept | One-liner |
|---|---|
| immutable | methods return new strings |
| `s[i]`, `s[-i]` | positive / negative indexing |
| `s[a:b:c]` | b excluded; `[::-1]` reverses |
| `find` vs `index` | -1 vs ValueError |
| `split`/`join` | str ⇄ list |
| f-string | `f'{x:.2f}'` — the modern default |
