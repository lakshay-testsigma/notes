## 1. Arithmetic Operators

| Op | Name | Example | Result |
|---|---|---|---|
| `+` | add | `5 + 2` | `7` |
| `-` | subtract | `5 - 2` | `3` |
| `*` | multiply | `5 * 2` | `10` |
| `/` | true divide | `5 / 2` | `2.5` (always float) |
| `//` | floor divide | `5 // 2` | `2` |
| `%` | modulo (remainder) | `5 % 2` | `1` |
| `**` | power | `5 ** 2` | `25` |

````python
print(5 + 2)
print(5 - 2)
print(5 * 2)
print(5 / 2)     # 2.5 — true division always gives float
print(5 // 2)    # 2   — floor division drops the fraction
print(5 % 2)     # 1   — remainder
print(5 ** 2)    # 25  — power
````

````python
# floor division floors toward NEGATIVE infinity — watch negatives
print(-5 // 2)   # -3, not -2
print(-5 % 2)    # 1 — % takes the sign of the DIVISOR (so a//b*b + a%b == a)
print(5 % -2)    # -1
````

````python
# math helpers that pair with the operators
print(abs(-7))         # magnitude
print(divmod(17, 5))   # (quotient, remainder) in one shot
print(pow(2, 10))      # same as 2 ** 10
print(pow(2, 10, 7))   # (2**10) % 7 — efficient modular power
````

## 2. Relational (Comparison) Operators

Always produce a `bool`.

````python
print(4 > 5)
print(4 < 5)
print(4 >= 4)
print(4 <= 3)
print(4 == 4)
print(4 != 4)
````

````python
# comparisons CHAIN — read like math
x = 7
print(1 < x < 10)          # same as (1 < x) and (x < 10)
print(10 <= x <= 20)
````

## 3. Logical Operators

Truth table (`x and y`, `x or y`, `not x`):

| x | y | and | or |
|---|---|-----|----|
| T | T | T | T |
| T | F | F | T |
| F | T | F | T |
| F | F | F | F |

````python
print(True and True)
print(True and False)
print(True or False)
print(not True)
````

````python
# 1 and 0 behave as True and False
print(1 and 0)   # 0
print(1 or 0)    # 1
print(not 1)     # False
````

### Truthiness — what counts as False?

| Falsy | Truthy |
|---|---|
| `False`, `None` | everything else |
| `0`, `0.0`, `0j` | any non-zero number |
| `''`, `[]`, `()`, `{}`, `set()` | any non-empty container |

````python
for value in [0, 1, '', 'hi', [], [0], None]:
    print(repr(value), '→', bool(value))
````

````python
# short-circuit: and/or return an OPERAND, not just True/False
name = ''
print(name or 'anonymous')     # fallback default idiom
print('admin' and 'logged in') # last truthy wins
````

## 4. Bitwise Operators

Work on the binary representation, bit by bit.

```
  2 = 0 1 0
  3 = 0 1 1
      ─────
  &   0 1 0  = 2     (1 only if BOTH bits 1)
  |   0 1 1  = 3     (1 if EITHER bit 1)
  ^   0 0 1  = 1     (1 if bits DIFFER)
```

````python
print(2 & 3)    # AND → 2
print(2 | 3)    # OR  → 3
print(2 ^ 3)    # XOR → 1
````

````python
# NOT: ~x == -(x+1)  (two's complement)
print(~3)       # -4
````

````python
# shifts: << multiplies by 2 per step, >> divides by 2 per step
print(4 >> 2)   # 0100 → 0001 = 1
print(4 << 2)   # 0100 → 10000 = 16
````

## 5. Assignment Operators

`=` assigns; compound forms modify in place: `+= -= *= /= //= %= **=`

⚠️ Python has **no `++` / `--`**.

````python
a = 3
a += 3
print(a)     # 6
a %= 4
print(a)     # 2
a **= 3
print(a)     # 8
````

````python
# ❌ no increment operator in Python → SyntaxError
a = 1
a++
````

## 6. Membership Operators

`in` / `not in` — test presence in a sequence or collection.

````python
print('D' in 'Delhi')
print('d' not in 'Delhi')     # case-sensitive!
print(3 in [1, 2, 3, 4])
print(5 not in {1, 2, 3})
````

## 6b. Identity Operators — `is` / `is not`

| | asks |
|---|---|
| `==` | same **value**? |
| `is` | same **object** (same `id`)? |

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b, a is b)    # equal values, different objects
print(a == c, a is c)    # same object — c is an alias
````

````python
# the ONE blessed use: comparing with None
x = None
print(x is None)
print(x is not None)

# ⚠️ never use `is` for numbers/strings — small ints are cached, so it
# sometimes 'works' by accident and breaks elsewhere. Use == for values.
````

### Program — sum of digits of a 3-digit number

Classic `%` / `//` combo: `% 10` grabs the last digit, `// 10` chops it off.

```
 345 % 10 = 5      345 // 10 = 34
  34 % 10 = 4       34 // 10 = 3
   3 % 10 = 3        3 // 10 = 0   stop
```

````python
# number = int(input('enter a 3 digit number: '))
number = 345

a = number % 10          # 5
number = number // 10    # 34
b = number % 10          # 4
c = number // 10         # 3

print(a + b + c)
````

## 7. If / Elif / Else

Indentation (4 spaces) **is** the block syntax — no braces.

````python
# login program
# email = input('enter email: ')
# password = input('enter password: ')
email = 'lakshay@example.com'
password = '1234'

if email == 'lakshay@example.com' and password == '1234':
    print('Welcome!')
elif email == 'lakshay@example.com' and password != '1234':
    print('Incorrect password')
else:
    print('Not registered')
````

````python
# minimum of 3 numbers
a, b, c = 12, 5, 9

if a < b and a < c:
    print('smallest:', a)
elif b < c:
    print('smallest:', b)
else:
    print('smallest:', c)
````

````python
# nested if — categorize a number
num = -7

if num >= 0:
    if num == 0:
        print('zero')
    else:
        print('positive')
else:
    print('negative')
````

### Conditional expression (ternary)

`value_if_true if condition else value_if_false` — an *expression*, not a statement.

````python
a, b = 4, 9
smaller = a if a < b else b
print(smaller)
````

## 8. `match-case` (Python 3.10+)

Structural pattern matching — a cleaner switch for multi-way branches.
`_` is the wildcard (default) case.

````python
def http_status(code):
    match code:
        case 200:
            return 'OK'
        case 301 | 302:            # multiple values with |
            return 'Redirect'
        case 404:
            return 'Not Found'
        case code if code >= 500:  # guard condition
            return 'Server Error'
        case _:
            return 'Unknown'

print(http_status(200))
print(http_status(302))
print(http_status(503))
print(http_status(42))
````

### Operator precedence (high → low, the ones that matter)

| Level | Operators |
|---|---|
| 1 | `**` |
| 2 | `~` unary `+ -` |
| 3 | `* / // %` |
| 4 | `+ -` |
| 5 | `<< >>` |
| 6 | `&` then `^` then `\|` |
| 7 | comparisons, `in`, `is` |
| 8 | `not` → `and` → `or` |

When in doubt: **use parentheses**.

````python
print(2 + 3 * 4 ** 2)      # 50, not 400
print((2 + 3) * 4 ** 2)    # 80
````
