# 05 · Lists

**TL;DR** — a list is a **mutable, ordered sequence of references** to any objects.

## How lists are stored in memory

A list doesn't hold values — it holds **pointers** to objects living elsewhere:

```
  L = [1, 'hi', [2, 3]]

  L ──► +--------+--------+--------+
        |  ref 0 |  ref 1 |  ref 2 |     (contiguous array of references)
        +---┬----+---┬----+---┬----+
            ▼        ▼        ▼
          int 1   str 'hi'  list [2,3]   (objects anywhere on the heap)
```

That's why lists can be heterogeneous, and why two names can share one list (aliasing).

**Agenda**: lists vs arrays → create → access → edit → delete → operations →
methods → comprehension → zip/enumerate → practice programs

## 1. Lists vs Arrays (C/Java-style)

| | Array | Python list |
|---|---|---|
| Size | fixed | dynamic (grows/shrinks) |
| Types | homogeneous | heterogeneous |
| Stores | values | references |
| Speed | faster | slower |
| Memory | compact | more overhead |

````python
# proof of reference storage: the SAME int object is reused
L = [1, 2, 3]
print(id(L[0]), id(1))   # same id — L[0] points at the shared int 1
````

### Characteristics
ordered · mutable · heterogeneous · duplicates allowed · dynamic · nestable · holds any object

````python
# order matters for equality
print([1, 2, 3] == [3, 2, 1])
````

## 3. Accessing — index & slice (same rules as strings)

````python
L = [1, 2, 3, 4, 5]
print(L[0], L[-1])
print(L[1:4])
print(L[::-1])
````

````python
# nested indexing — chain the brackets
matrix = [[1, 2], [3, 4]]
print(matrix[1][0])          # 3

deep = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
print(deep[1][0][1])         # 6
````

## 4. Adding Items — `append`, `extend`, `insert`

````python
L = [1, 2, 3]
L.append(4)          # one item at the end
print(L)

L.append([5, 6])     # ⚠️ appends the LIST as one item
print(L)
````

````python
L = [1, 2, 3]
L.extend([4, 5, 6])  # spreads items in
print(L)

L.extend('hi')       # a string is iterable → chars spread in!
print(L)
````

````python
L = [1, 2, 4]
L.insert(2, 3)       # insert(position, value)
print(L)
````

## 5. Editing — lists ARE mutable

````python
L = [1, 2, 3, 4, 5]
L[-1] = 500          # edit one item
print(L)

L[1:4] = [200, 300, 400]   # edit a whole slice
print(L)
````

## 6. Deleting — `del`, `remove`, `pop`, `clear`

| | deletes by | returns |
|---|---|---|
| `del L[i]` | index / slice | — |
| `remove(v)` | first matching **value** | — |
| `pop(i)` | index (default last) | the removed item |
| `clear()` | everything | — |

````python
L = [1, 2, 3, 4, 5]
del L[-1]
print(L)
del L[1:3]
print(L)
````

````python
L = [10, 20, 30, 20]
L.remove(20)        # only the FIRST 20
print(L)

last = L.pop()
print(last, L)

L.clear()
print(L)
````

## 7. Operations

````python
L1, L2 = [1, 2, 3], [4, 5, 6]
print(L1 + L2)       # concat (new list)
print(L1 * 2)        # repeat
````

````python
# membership — nested lists match as whole items
L2 = [1, 2, 3, [5, 6]]
print(5 in L2)         # False — 5 is inside the nested list
print([5, 6] in L2)    # True
````

````python
# loops
for item in [[1, 2], [3, 4]]:
    print(item)
````

## 8. Functions & Methods

````python
L = [2, 1, 5, 7, 0]
print(len(L), min(L), max(L))
print(sum(L), sum(L, 100))        # sum, and sum with a start value
print(sorted(L))                  # NEW sorted list
print(sorted(L, reverse=True))
print(L)                          # unchanged
````

````python
# key= — sort/pick by a computed value, not the item itself
words = ['kiwi', 'fig', 'banana']
print(sorted(words, key=len))
print(max(words, key=len))
words.sort(key=len, reverse=True)   # .sort takes key too
print(words)
````

````python
L = [2, 1, 5, 7, 0]
L.sort()                          # in-place, returns None
print(L)

L.reverse()                       # in-place reverse
print(L)
````

````python
# reversed() — lazy view, original untouched (pairs with sorted vs sort)
L = [1, 2, 3]
print(list(reversed(L)), L)
````

````python
L = [1, 2, 1, 3, 1]
print(L.count(1))
print(L.index(3))                 # first position of value
````

### Aliasing vs Copy

```
 aliasing:  L ──►┌────────┐◄── L1     (one list, two names)
                 │  list  │
 copy:      L ──►└────────┘
            L1 ──►┌────────┐          (two separate lists)
                  │ clone  │
```

````python
L = [1, 2, 3]
alias = L            # NO copy — same object
alias.append(4)
print(L)             # L changed too!
print(id(L) == id(alias))
````

````python
L = [1, 2, 3]
clone = L.copy()     # shallow copy — new outer list
clone.append(4)
print(L, clone)
print(id(L) == id(clone))
````

````python
# shallow copy shares NESTED objects — deepcopy clones all levels
import copy

L = [[1, 2], [3, 4]]
shallow = L.copy()
deep = copy.deepcopy(L)

L[0].append(99)
print('shallow sees it:', shallow[0])   # [1, 2, 99]
print('deep does not:  ', deep[0])      # [1, 2]
````

## 9. List Comprehension

`[expression for item in iterable if condition]` — build a list in one line.

Why prefer it over a for-append loop? Faster (no repeated `.append` lookups),
fewer lines, and it reads as a formula ("what the list IS"), not a recipe.

```
 [ x**2   for x in range(5)   if x % 2 == 0 ]
   ─────      ─────────────      ───────────
   output       source             filter
```

````python
print([i for i in range(1, 11)])
print([i * 2 for i in [1, 2, 3, 4]])       # scalar multiply
print([i ** 2 for i in range(1, 6)])       # squares
````

````python
# with filter
print([i for i in range(1, 51) if i % 5 == 0])
````

````python
# if/else in the OUTPUT position (before the for) — transform, don't drop
print(['even' if i % 2 == 0 else 'odd' for i in range(1, 8)])
````

````python
# multiple ifs chain — every filter must pass (an implicit and)
basket = ['apple', 'guava', 'cherry', 'avocado']
my_fruits = ['apple', 'kiwi', 'avocado', 'banana']
print([f for f in my_fruits if f in basket if f.startswith('a')])
````

````python
# filter on strings
languages = ['java', 'python', 'php', 'c', 'javascript']
print([lang for lang in languages if lang.startswith('p')])
````

````python
# nested comprehension — 3x3 multiplication matrix
print([[i * j for j in range(1, 4)] for i in range(1, 4)])
````

````python
# cartesian product — two fors
print([i * j for i in [1, 2, 3, 4] for j in [5, 6, 7, 8]])
````

## 10. Traversal Idioms — itemwise, indexwise, `enumerate`, `zip`

````python
L = ['a', 'b', 'c']

# itemwise
for item in L:
    print(item)

# indexwise
for i in range(len(L)):
    print(i, L[i])
````

````python
# enumerate — index AND item, no bookkeeping
for i, item in enumerate(L):
    print(i, item)

# human-friendly numbering: start=1
for rank, item in enumerate(L, start=1):
    print(rank, item)
````

````python
# zip — walk two lists in lockstep; itemwise addition
L1, L2 = [1, 2, 3, 4], [-1, -2, -3, -4]
print(list(zip(L1, L2)))
print([a + b for a, b in zip(L1, L2)])
````

````python
# unequal lengths: the SHORTEST iterable decides — extras are dropped
print(list(zip([1, 2, 3, 4, 5], ['a', 'b', 'c'])))
````

````python
# zip(*rows) — unzip / transpose a matrix
rows = [(1, 2, 3), (4, 5, 6)]
print(list(zip(*rows)))     # columns: [(1, 4), (2, 5), (3, 6)]
````

````python
# lists can hold ANYTHING — even functions
funcs = [len, max, min]
for f in funcs:
    print(f([3, 1, 4, 1, 5]))
````

### Disadvantages of lists
- slower than typed arrays (numpy) for math
- reference indirection costs memory
- aliasing bugs — two names silently sharing one list

## 11. Practice Programs

````python
# split into odds and evens
L = [1, 2, 3, 4, 5, 6]
odds = [x for x in L if x % 2]
evens = [x for x in L if x % 2 == 0]
print(odds, evens)
````

````python
# take a list as input from the user (comma-separated)
# raw = input('enter numbers separated by comma: ')
raw = '10,20,30,40'
L = [int(x) for x in raw.split(',')]
print(L, type(L))
````

````python
# merge two lists without +
L1, L2 = [1, 2, 3], [4, 5, 6]
merged = L1.copy()
merged.extend(L2)
print(merged)
````

````python
# replace an item if found
L = [1, 2, 3, 4, 3]
old, new = 3, 300
for i in range(len(L)):
    if L[i] == old:
        L[i] = new
print(L)
````

````python
# 2D list → 1D (flatten)
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [x for row in matrix for x in row]
print(flat)
````

````python
# remove duplicates, keep order
L = [1, 2, 1, 3, 2, 4]
unique = []
for x in L:
    if x not in unique:
        unique.append(x)
print(unique)
````

````python
# is the list ascending?
L = [1, 2, 2, 5, 9]
ascending = True
for i in range(len(L) - 1):
    if L[i] > L[i + 1]:
        ascending = False
        break
print(ascending)
````

````python
# same, the idiomatic way
print(L == sorted(L))
````
