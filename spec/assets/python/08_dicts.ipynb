# 08 · Dictionaries

**TL;DR** — a dict maps **unique, hashable keys → values**. Python's built-in
hash table / associative array; the shape behind JSON.

```
           hash(key)          bucket
  'name' ────────────► 3 ──► ('name', 'lakshay')
  'age'  ────────────► 7 ──► ('age', 25)

  lookup by key ≈ O(1) — no scanning
```

**Agenda**: create → access → add/edit → delete → iterate → views →
merge → dict comprehension

## 1. Creating Dicts

Characteristics: mutable · keys unique · keys must be hashable · no positional indexing

````python
empty = {}
person = {'name': 'lakshay', 'age': 25}
mixed_keys = {(1, 2): 'tuple key', 3: 'int key', 'four': 'str key'}
from_pairs = dict([('name', 'lakshay'), ('age', 25)])

print(empty, person, mixed_keys, from_pairs, sep='\n')
````

````python
# duplicate keys: the LAST value silently wins
d = {'name': 'lakshay', 'name': 'goyal'}
print(d)
````

````python
# ❌ mutable (unhashable) key → TypeError
d = {[1, 2]: 'list key'}
````

````python
# nested dict — JSON-style
student = {
    'name': 'lakshay',
    'marks': {'maths': 90, 'english': 85},
    'friends': ['ravi', 'asha'],
}
print(student)
````

## 2. Accessing Values — `[]` vs `.get()`

| | key exists | key missing |
|---|---|---|
| `d[key]` | value | ❌ KeyError |
| `d.get(key)` | value | `None` (or your default) |

````python
person = {'name': 'lakshay', 'age': 25}
print(person['name'])
print(person.get('age'))
print(person.get('salary'))            # None, no crash
print(person.get('salary', 0))         # custom default
````

````python
# ❌ missing key with [] → KeyError
person = {'name': 'lakshay'}
person['salary']
````

````python
# nested access — chain the keys
student = {'name': 'lakshay', 'marks': {'maths': 90, 'english': 85}}
print(student['marks']['maths'])
````

## 3. Adding & Editing — same syntax

Assign to a key: new key → added, existing key → updated.

````python
d = {'name': 'lakshay'}
d['age'] = 25              # add
d['name'] = 'goyal'        # edit
print(d)
````

````python
# nested add
student = {'name': 'lakshay', 'marks': {'maths': 90}}
student['marks']['science'] = 88
print(student)
````

## 4. Removing — `pop`, `popitem`, `del`, `clear`

````python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4}

print(d.pop('b'))      # remove by key, returns value
print(d.popitem())     # removes LAST-inserted pair
del d['a']
print(d)

d.clear()
print(d)
````

````python
# nested delete
student = {'name': 'lakshay', 'marks': {'maths': 90, 'english': 85}}
del student['marks']['english']
print(student)
````

## 5. Membership & Iteration

⚠️ Both check/walk **keys** by default.

````python
person = {'name': 'lakshay', 'age': 25}
print('name' in person)      # key → True
print('lakshay' in person)   # value → False
````

````python
for key in person:
    print(key, '→', person[key])
````

## 6. Views & Methods — `keys` / `values` / `items` / `update`

````python
person = {'name': 'lakshay', 'age': 25}
print(person.keys())
print(person.values())
print(person.items())
````

````python
# items() unpacks beautifully in loops
for key, value in person.items():
    print(key, '→', value)
````

````python
# update — bulk add/overwrite
d = {'a': 1, 'b': 2}
d.update({'b': 20, 'c': 3})
print(d)
````

````python
# setdefault — get, but INSERT the default if the key is missing
groups = {}
for word in ['apple', 'ant', 'bat', 'bee', 'axe']:
    groups.setdefault(word[0], []).append(word)
print(groups)
````

````python
# fromkeys — same starting value for many keys
print(dict.fromkeys(['mon', 'tue', 'wed'], 0))
````

````python
# merge operators (3.9+): | new dict, |= in place
d1 = {'a': 1, 'b': 2}
d2 = {'b': 20, 'c': 3}
print(d1 | d2)         # right side wins ties
d1 |= d2
print(d1)
````

````python
# functions on dicts
d = {'x': 3, 'y': 1, 'z': 2}
print(len(d))
print(sorted(d))               # sorted keys
print(max(d))                  # max key
print(sorted(d, key=d.get))    # keys sorted BY VALUE
````

## 7. Dict Comprehension

`{key_expr: value_expr for item in iterable if condition}`

````python
# squares
print({i: i ** 2 for i in range(1, 11)})
````

````python
# transform an existing dict via .items()
distances_km = {'delhi': 1000, 'mumbai': 2000, 'bangalore': 3000}
print({city: km * 0.62 for city, km in distances_km.items()})   # → miles
````

````python
# build from two lists via zip
days = ['sun', 'mon', 'tue']
temps = [30.5, 32.6, 31.8]
print({day: temp for day, temp in zip(days, temps)})
````

````python
# with a filter
products = {'phone': 10, 'laptop': 0, 'charger': 32, 'tablet': 0}
print({item: qty for item, qty in products.items() if qty > 0})
````

````python
# nested comprehension — multiplication tables
print({i: {j: i * j for j in range(1, 6)} for i in range(2, 5)})
````

---
## Recap

| Concept | One-liner |
|---|---|
| keys | unique + hashable; last duplicate wins |
| `[]` vs `get` | KeyError vs None/default |
| add = edit | `d[k] = v` does both |
| iteration | keys by default; `.items()` for pairs |
| merge | `d1 \| d2` new, `d1 \|= d2` in place |
| comprehension | `{k: v for ... if ...}` |
