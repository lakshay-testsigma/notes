# 07 · Sets

**TL;DR** — a set is an **unordered bag of unique, hashable items**. Built for
membership tests and math-style set algebra.

```
    A = {1,2,3,4}        B = {3,4,5,6}

        ┌───────A───────┐
        │  1  2  ┌──────┼──────B──────┐
        │        │ 3  4 │  5  6       │
        └────────┼──────┘             │
                 └────────────────────┘
  A | B  = {1,2,3,4,5,6}   union
  A & B  = {3,4}           intersection
  A - B  = {1,2}           difference
  A ^ B  = {1,2,5,6}       symmetric difference
```

**Agenda**: create → what's banned → add/remove → set algebra → subset tests →
frozenset → set comprehension

## 1. Creating Sets

Characteristics: unordered · mutable · **no duplicates** · items must be **hashable**
(immutable: int, str, tuple — not list/dict/set)

````python
empty = set()            # ⚠️ {} makes a DICT, not a set
print(type(empty), type({}))

s1 = {1, 2, 3}
hetero = {1, 'hello', 4.5, (1, 2)}
from_list = set([1, 2, 2, 3, 3, 3])    # duplicates collapse
print(s1, hetero, from_list)
````

````python
# ❌ mutable items not allowed → TypeError: unhashable type: 'list'
s = {1, 2, [3, 4]}
````

````python
# order is NOT preserved; equality ignores order
s1 = {1, 2, 3}
s2 = {3, 2, 1}
print(s1 == s2)
````

## 2. What sets DON'T support

No order → no indexing, no slicing, no item editing.

````python
# ❌ indexing → TypeError
s = {1, 2, 3}
s[0]
````

````python
# ❌ slicing → TypeError
s = {1, 2, 3}
s[0:2]
````

````python
# ❌ del by index → TypeError (del s — the whole set — is fine)
s = {1, 2, 3}
del s[0]
````

## 3. Adding & Removing

````python
s = {1, 2, 3}
s.add(4)             # one item
s.update([5, 6])     # many items
print(s)
````

````python
# remove vs discard: both delete by value, but...
s = {1, 2, 3, 4}
s.remove(4)          # ❌ raises KeyError if missing
s.discard(99)        # ✅ silent if missing
print(s)
````

````python
# ❌ remove() on a missing value → KeyError
s = {1, 2, 3}
s.remove(99)
````

````python
s = {1, 2, 3}
popped = s.pop()     # removes an ARBITRARY item
print(popped, s)

s.clear()
print(s)

del s                # unbind the name entirely
````

## 4. Set Algebra

| Operation | Operator | Method | In-place method |
|---|---|---|---|
| union | `A \| B` | `A.union(B)` | `A.update(B)` |
| intersection | `A & B` | `A.intersection(B)` | `A.intersection_update(B)` |
| difference | `A - B` | `A.difference(B)` | `A.difference_update(B)` |
| symmetric diff | `A ^ B` | `A.symmetric_difference(B)` | `A.symmetric_difference_update(B)` |

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)     # union
print(A & B)     # intersection
print(A - B)     # in A, not in B
print(B - A)
print(A ^ B)     # in exactly one of them
````

````python
# method forms return NEW sets...
print(A.union(B))
print(A.intersection(B))
print(A)                     # A unchanged
````

````python
# ..._update forms modify IN PLACE
A = {1, 2, 3, 4}
A.intersection_update(B)
print(A)                     # A is now {3, 4}
````

## 5. Relationship Tests

````python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {7, 8}

print(A.issubset(B))       # every A-item in B?
print(B.issuperset(A))
print(A.isdisjoint(C))     # nothing in common?
````

````python
# membership & iteration & functions
s = {4, 1, 3, 2}
print(3 in s)
print(len(s), sum(s), min(s), max(s))
print(sorted(s))           # sorted LIST from a set
````

````python
# copy
s = {1, 2, 3}
s2 = s.copy()
s2.add(4)
print(s, s2)
````

## 6. Frozenset — the immutable set

Read operations work; anything that writes does not exist.

````python
fs = frozenset([1, 2, 3, 4])

# ✅ works — all READ operations
print(fs | frozenset([5]))
print(fs & frozenset([3, 4, 5]))
print(3 in fs)
print(len(fs), sum(fs), max(fs))
print(fs.issubset({1, 2, 3, 4, 5}))
````

````python
# ❌ write operations don't exist → AttributeError
fs = frozenset([1, 2, 3])
fs.add(4)
````

````python
# being hashable, frozensets can nest inside sets → '2D sets'
nested = {frozenset([1, 2]), frozenset([3, 4])}
print(nested)
````

## 7. Set Comprehension

````python
print({i ** 2 for i in range(1, 11) if i % 2 == 0})
````

````python
# practical: unique word lengths in a sentence
sentence = 'the quick brown fox jumps over the lazy dog'
print({len(word) for word in sentence.split()})
````
