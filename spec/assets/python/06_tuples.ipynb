# 06 · Tuples

**TL;DR** — a tuple is an **immutable list**: ordered, indexable, but frozen at birth.

```
  list  [1, 2, 3]   ──► editable   (append/remove/sort...)
  tuple (1, 2, 3)   ──► read-only  (index/slice/count only)
```

**Agenda**: create → access → why immutable → operations → list vs tuple →
unpacking → zip

## 1. Creating Tuples

````python
empty = ()
homo = (1, 2, 3, 4)
hetero = (1, True, 5.6, 'hello')
nested = (1, 2, (3, 4))
from_str = tuple('hello')

print(empty, homo, hetero, nested, from_str, sep='\n')
````

````python
# ⚠️ single item NEEDS the trailing comma
not_a_tuple = (2)
real_tuple = (2,)
print(type(not_a_tuple), type(real_tuple))
````

````python
# parentheses are optional — the COMMA makes the tuple
t = 1, 2, 3
print(t, type(t))
````

## 2. Accessing — exactly like lists

````python
t = (10, 20, 30, 40, 50)
print(t[0], t[-1])
print(t[1:4])
print(t[::-1])

nested = (1, 2, (3, 4))
print(nested[2][1])
````

## 3. Immutability — what fails

````python
# ❌ editing → TypeError
t = (1, 2, 3)
t[0] = 100
````

````python
# ❌ appending — tuples have no append → AttributeError
t = (1, 2, 3)
t.append(4)
````

````python
# ❌ deleting one item → TypeError
t = (1, 2, 3)
del t[0]
````

````python
# deleting the WHOLE tuple works (unbinds the name)
t = (1, 2, 3)
del t
````

````python
# ⚠️ immutable ≠ frozen contents: a MUTABLE item inside can still change
t = (1, 2, [3, 4])
t[2].append(5)      # allowed — the list object mutated, tuple refs unchanged
print(t)
````

## 4. Operations & Methods

````python
t1, t2 = (1, 2, 3), (4, 5, 6)
print(t1 + t2)
print(t1 * 3)
print(1 in t1)

for item in t1:
    print(item, end=' ')
print()
````

````python
t = (2, 7, 1, 7, 5)
print(len(t), sum(t), min(t), max(t))
print(sorted(t))           # returns a LIST
print(t.count(7))
print(t.index(5))
print(t.count(50))         # missing value → 0, no error
````

````python
# ❌ index() on a missing value → ValueError (same rule as strings/lists)
(2, 7, 1).index(50)
````

## 5. Why use a tuple at all?

1. **Speed** — creation & iteration faster than list
2. **Memory** — smaller footprint
3. **Safety** — can't be mutated by accident (constants, config)
4. **Dict keys / set members** — hashable (lists are not)
5. **Multiple return values** — functions return tuples naturally

````python
# speed: creating 10M literals
import time

start = time.time()
for _ in range(10_000_000):
    x = [1, 2, 3, 4, 5]
list_time = time.time() - start

start = time.time()
for _ in range(10_000_000):
    x = (1, 2, 3, 4, 5)
tuple_time = time.time() - start

print(f'list : {list_time:.3f}s')
print(f'tuple: {tuple_time:.3f}s')
````

````python
# memory
import sys

L = list(range(1000))
T = tuple(range(1000))
print('list :', sys.getsizeof(L), 'bytes')
print('tuple:', sys.getsizeof(T), 'bytes')
````

````python
# hashable → usable as dict key; list is not
locations = {(28.6, 77.2): 'Delhi', (19.1, 72.9): 'Mumbai'}
print(locations[(28.6, 77.2)])
````

### List vs Tuple

| | list | tuple |
|---|---|---|
| Syntax | `[1, 2]` | `(1, 2)` |
| Mutable | ✅ | ❌ |
| Speed / memory | slower / bigger | faster / smaller |
| Methods | 11 (append, sort, ...) | 2 (count, index) |
| Dict key | ❌ | ✅ |
| Error-prone aliasing | ✅ risk | ❌ safe |

````python
# aliasing risk demo: tuple 'aliases' can never bite you
a = [1, 2, 3]
b = a
a.append(4)
print('list alias :', b)      # b changed too

a = (1, 2, 3)
b = a
a = a + (4,)                  # creates a NEW tuple, b untouched
print('tuple rebind:', b)
````

## 6. Tuple Unpacking

````python
a, b, c = (1, 2, 3)
print(a, b, c)
````

````python
# ❌ count mismatch → ValueError
a, b = (1, 2, 3)
````

````python
# swap — the Python classic, no temp variable
a, b = 1, 2
a, b = b, a
print(a, b)
````

````python
# star-unpacking — soak up the rest
a, b, *rest = (1, 2, 3, 4, 5)
print(a, b, rest)

first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
````

````python
# zip on tuples → tuple of pairs
t1, t2 = (1, 2, 3), (4, 5, 6)
print(tuple(zip(t1, t2)))
````
