# 03 · Loops

**TL;DR** — repeat work without repeating code.

```
       ┌──────────────┐
       ▼              │
  condition? ──True──►body
       │              │
     False            │
       ▼              │
     exit ◄───────────┘ (break jumps here early;
                         continue jumps back up)
```

**Agenda**: while → while-else → for → range → nested loops → patterns →
break/continue/pass → for-else → practice programs

## 1. `while` — repeat while a condition holds

Use when you **don't know in advance** how many iterations you need.

````python
# print the multiplication table of a number
number = 12
i = 1

while i <= 10:
    print(number, '*', i, '=', number * i)
    i += 1
````

### `while ... else`

The `else` block runs when the loop ends **normally** (condition became False) —
it is skipped if the loop exits via `break`.

````python
# guessing game — while...else
# In Colab, replace the preset guesses with: guess = int(input('guess: '))
import random
random.seed(1)
target = random.randint(1, 100)

guesses = [50, 25, 12, 19, target]   # simulated user guesses, halving the range
attempts = 0
i = 0

while guesses[i] != target:
    attempts += 1
    if guesses[i] < target:
        print(guesses[i], '→ guess higher')
    else:
        print(guesses[i], '→ guess lower')
    i += 1
else:
    print('correct! took', attempts + 1, 'tries; target was', target)
````

## 2. `for` — iterate over a sequence

Use when you have something to **walk through**: string, list, tuple, set, dict, range.

````python
for ch in 'Delhi':
    print(ch)
````

````python
for item in [1, 2, 3]:
    print(item)

for item in (4, 5, 6):
    print(item)

for item in {7, 8, 9}:      # set — order not guaranteed
    print(item)

for key in {'a': 1, 'b': 2}:  # dict — iterates keys
    print(key)
````

### `range(start, stop, step)` — lazy number sequence

`stop` is **excluded**.

````python
print(list(range(5)))            # 0..4
print(list(range(2, 8)))         # 2..7
print(list(range(1, 11, 2)))     # odd numbers
print(list(range(10, 0, -1)))    # countdown — negative step
````

### For vs While — when to use what?

| | `for` | `while` |
|---|---|---|
| Iterations known? | yes (sequence/range) | no (condition-driven) |
| Typical use | walk a collection | wait for event, converge on value |
| Risk | low | infinite loop if condition never flips |

## 3. Practice — condition-driven programs

````python
# population: currently 10000, grows 10% per year — print last 10 years' values
curr_pop = 10000

for i in range(10, 0, -1):
    print(i, 'years ago →', round(curr_pop))
    curr_pop /= 1.1
````

````python
# sum of series  1/1! + 2/2! + 3/3! + ... + n/n!
n = 10
result = 0
fact = 1

for i in range(1, n + 1):
    fact *= i
    result += i / fact

print(round(result, 6))
````

````python
# sum of all digits of a given number
# number = int(input('number: '))
number = 76234
total = 0

while number > 0:
    total += number % 10
    number //= 10

print(total)
````

````python
# keep accepting numbers until user enters 0, then print the average
# In Colab:  num = int(input('enter number (0 to stop): '))
entries = [12, 7, 41, 9, 0]   # simulated user input, 0 ends

total = 0
count = 0
i = 0
while entries[i] != 0:
    total += entries[i]
    count += 1
    i += 1

print('average:', total / count)
````

## 4. Nested Loops

A loop inside a loop: the inner loop completes fully for **each** outer iteration.

```
 outer i=1 ──► inner j=1..3
 outer i=2 ──► inner j=1..3     total = outer × inner iterations
```

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
````

````python
# unique combinations of 1,2,3,4 (order doesn't matter, no repeats)
for i in range(1, 5):
    for j in range(i + 1, 5):
        print(i, j)
````

### Pattern 1 — solid star triangle

```
*
**
***
****
```

````python
rows = 4
for i in range(1, rows + 1):
    print('*' * i)
````

````python
# same, built with a nested loop
rows = 4
for i in range(1, rows + 1):
    for j in range(i):
        print('*', end='')
    print()
````

### Pattern 2 — number palindrome pyramid

```
   1
  121
 12321
1234321
```

````python
rows = 4
for i in range(1, rows + 1):
    print(' ' * (rows - i), end='')
    for j in range(1, i + 1):        # ascending 1..i
        print(j, end='')
    for j in range(i - 1, 0, -1):    # descending i-1..1
        print(j, end='')
    print()
````

## 5. Loop Control — `break`, `continue`, `pass`

```
 break    ──► exit the loop entirely
 continue ──► skip rest of THIS iteration, go to next
 pass     ──► do nothing (placeholder)
```

````python
# break demo — stop at 5
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

````python
# continue demo — skip 5
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
````

````python
# pass demo — valid empty body, fill in later
for i in range(1, 5):
    pass
````

````python
# `while True` + break — loop until an inner condition says stop
# (the standard shape for menus and retry loops)
attempts = ['wrong', 'nope', 'secret']    # simulated input('password: ')
i = 0
while True:
    if attempts[i] == 'secret':
        print('unlocked after', i + 1, 'tries')
        break
    i += 1
````

### `for ... else` + `break` — the search idiom

`else` runs only if the loop finished **without** `break` → perfect for
"searched everything, found nothing".

````python
# prime numbers in a range (linear search with break)
lower, upper = 10, 40

for n in range(lower, upper + 1):
    if n < 2:
        continue
    for divisor in range(2, n):
        if n % divisor == 0:
            break                 # found a divisor → not prime
    else:
        print(n, 'is prime')      # no break happened → prime
````

````python
# continue example (e-commerce): apply discount only to eligible items
cart = [
    {'item': 'laptop',   'price': 60000, 'discountable': True},
    {'item': 'gift card','price': 2000,  'discountable': False},
    {'item': 'mouse',    'price': 900,   'discountable': True},
]

total = 0
for product in cart:
    if not product['discountable']:
        total += product['price']
        continue                        # skip discount logic
    total += product['price'] * 0.9     # 10% off

print('payable:', total)
````

## Recap

| Concept | One-liner |
|---|---|
| `while` | unknown iteration count |
| `for` + `range` | known count / walk a sequence |
| `range(a, b, s)` | b excluded, step may be negative |
| nested loops | inner completes per outer step |
| `break` | exit loop now |
| `continue` | skip to next iteration |
| `pass` | placeholder, does nothing |
| loop `else` | runs only if no `break` |
